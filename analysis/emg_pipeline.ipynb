```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/test_log.csv")
df.head()
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df["timestamp_ms"], df["value"])
plt.xlabel("Time (ms)")
plt.ylabel("ADC value")
plt.title("Raw EMG envelope - test recording")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/test_log.csv", header=None, names=["timestamp_us", "value"])
df.head()
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df["timestamp_us"], df["value"])
plt.xlabel("Time (us)")
plt.ylabel("ADC value")
plt.title("Raw EMG envelope - test recording")
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("../data/raw/test_log.csv", header=None, names=["timestamp_us", "value"])
deltas = df["timestamp_us"].diff()

print(deltas.describe())
print("Effective rate (Hz):", 1e6 / deltas.mean())
```
